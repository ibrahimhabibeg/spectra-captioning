```python
def get_qualitative_distance(z: float) -> str:
    """Return a qualitative description of distance based on redshift."""
    if z < 0.1:
        return "very close (local universe)"
    elif z < 0.5:
        return "close"
    elif z < 1.0:
        return "intermediate"
    elif z < 2.0:
        return "far"
    else:
        return "very far (high redshift)"
    
df['distance_description'] = df['Z'].apply(get_qualitative_distance)
df['distance_description'].value_counts(normalize=True)
```
